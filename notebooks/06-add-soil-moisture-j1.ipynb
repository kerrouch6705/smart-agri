### 01 - Ajout de la cible Soil_Moisture_J1

Dans ce notebook, nous ajoutons une nouvelle colonne appelée `Soil_Moisture_J1`.

Cette colonne représente une estimation de l'humidité du sol au jour suivant.



```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv(r"C:\Users\kerro\OneDrive\Bureau\smart-agri\data\processed\irrigation_prediction_18cols.csv")

df.head(10)
```

## Création de la colonne Soil_Moisture_J1

L'humidité du sol au jour suivant peut être influencée par plusieurs facteurs :

- la pluie augmente l'humidité du sol,
- l'irrigation précédente augmente l'humidité du sol,
- l'humidité de l'air peut contribuer à garder le sol humide,
- la température élevée diminue l'humidité,
- le vent diminue l'humidité,
- les heures d'ensoleillement diminuent l'humidité,
- un indice de stress élevé indique une baisse de l'humidité.

```python
df["Soil_Moisture_J1"] = (
    df["Soil_Moisture"]
    + 0.015 * df["Rainfall_mm"]
    + 0.20 * df["Previous_Irrigation_mm"]
    + 0.05 * df["Humidity"]
    - 0.30 * df["Temperature_C"]
    - 0.40 * df["Wind_Speed_kmh"]
    - 0.50 * df["Sunlight_Hours"]
    - 0.10 * df["Stress_Index"]
)
```

```python
df["Soil_Moisture_J1"] = df["Soil_Moisture_J1"].clip(0, 100)

df[["Soil_Moisture", "Soil_Moisture_J1"]].head(15)
```

```python
df.to_csv("../data/processed/irrigation_prediction_cols19.csv", index=False)
```
